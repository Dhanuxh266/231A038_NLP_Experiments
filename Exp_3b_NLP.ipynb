```python
# Import Libraries & Download Resources
import string
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import re
import emoji

nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# Original Text
print("Dhanush Pallan — 231A038")
text = '''Hey everyone!!! 👋

I'm SUPER excited 😊 because I've successfully completed my first Machine Learning project on 25th July, 2026. The project analyzed more than 12,800 customer reviews collected from https://www.example.com and an additional dataset downloaded from www.kaggle.com.

Unfortunately, some reviews were not very helpful—they contained only words like "the", "is", "and", "of", "to", etc. Others included unusual symbols such as @#$%^&*, repeated punctuation marks!!!!!!, HTML tags like `<br>`, and emojis 😄😄😄.

You can contact me at vijay.shanker@eng.rizvi.edu.in or support123@company.co.in if you're interested in getting the dataset. Follow us on Twitter @AI_ResearchLab and use hashtags such as #MachineLearning #NLP #DataScience.

The model's accuracy improved from 83.2% to 92.1%, while the training time decreased by approximately 34%. It processed 1,28,000 records in under 2.5 hours and used around 3.8 GB of RAM.

Students were learning, analyzing, processing, predicting, running, writing, and comparing various algorithms. The researchers identified improved techniques after conducting several experiments and evaluating different models.

Don't feel discouraged if the results aren't perfect—every experiment is a learning opportunity! Let's make them better together. I'll share the final report by 02/08/2026.

Thank you so much!!! 🙌'''
text
```

```python
# Text to Lowercase
print("Dhanush Pallan — 231A038")
text_lower = text.lower()
text_lower
```

```python
# Removing URLs and emails, Twitter mentions and hashtags, and numbers
print("Dhanush Pallan — 231A038")
# Initialize processed_text with the first cleaning step
processed_text = re.sub(r'http\S+|www\.\S+', '', text_lower)
# Apply subsequent cleaning steps to the already processed text
processed_text = re.sub(r'\S+@\S+', '', processed_text)
processed_text = re.sub(r'@\S+', '', processed_text)
processed_text = re.sub(r'#\S+', '', processed_text)
# Remove numbers and any immediately attached non-whitespace characters
processed_text = re.sub(r'\d+\S*', '', processed_text)
# Normalize spaces after all removals
text_no_url = re.sub(r'\s+', ' ', processed_text)
text_no_url
```

```python
# Removing all emojis
print("Dhanush Pallan — 231A038")
text_no_emoji = emoji.replace_emoji(text_no_url, replace='')
text_no_emoji
```

```python
print("Dhanush Pallan — 231A038")
punctuations = list(string.punctuation)
print(punctuations)
```

```python
# Tokenized Text
print("Dhanush Pallan — 231A038")
tokenized_text = word_tokenize(text_no_emoji)
print(tokenized_text)
```

```python
# Removing Punctuation
print("Dhanush Pallan — 231A038")
cleaned_text1 = []
for word in tokenized_text:
  if word not in punctuations:
    cleaned_text1.append(word)
print(cleaned_text1)
print(len(cleaned_text1))
```

```python
print("Dhanush Pallan — 231A038")
print(stopwords.words('english'))
print(len(stopwords.words('english')))
```

```python
# Removing Stopwords
print("Dhanush Pallan — 231A038")
cleaned_text2 = []
for word in cleaned_text1:
  if word not in stopwords.words('english'):
    cleaned_text2.append(word)
print(cleaned_text2)
print(len(cleaned_text2))
```

```python
# Stemming
print("Dhanush Pallan — 231A038")
stemmer = PorterStemmer()
stemmed_text = []
for word in cleaned_text2:
  stemmed_text.append(stemmer.stem(word))
print(stemmed_text)
```

```python
# Lemmatization
print("Dhanush Pallan — 231A038")
lemmatizer = WordNetLemmatizer()
lemmatized_text = []
for word in cleaned_text2:
  lemmatized_text.append(lemmatizer.lemmatize(word))
print(lemmatized_text)
```

```python
print("Dhanush Pallan — 231A038")
print("Original Text:")
print(text)
print("\nStemmed Text:")
print(stemmed_text)
print("Doug was here...")
print("\nLemmatized Text:")
print(lemmatized_text)
```
