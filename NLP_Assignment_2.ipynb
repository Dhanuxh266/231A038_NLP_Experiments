**Q.1 Given the sentence: “The bank will not open until the river recedes.”**

(a) Identify and explain the type of lexical ambiguity present.
The word “bank” exhibits lexical ambiguity, specifically polysemy or homonymy, depending on the relationship between its meanings.
- Financial bank: An institution that accepts deposits and provides financial services.
- River bank: The land alongside a river.
In this sentence, the word “bank” refers to the land alongside the river. The phrase “until the river recedes” provides the contextual clue that helps identify the intended meaning.
(b) Suggest how WordNet or BabelNet could help resolve this ambiguity.
WordNet is a lexical database that groups words into sets of synonyms called synsets, with each synset representing a distinct sense.

It could help resolve the ambiguity as follows:
1. Identify the ambiguous word “bank.”
2. Retrieve its possible senses from WordNet, including a financial institution and the land bordering a river.
3. Examine the surrounding words, such as “river” and “recedes.”
4. Select the river-edge sense because it is semantically consistent with the context.
BabelNet combines multilingual lexical resources and encyclopedic knowledge. It can provide additional semantic information and links between concepts across languages, helping a system identify the appropriate sense.
(c) Explain why automatic systems might still fail in some contexts.
Automatic systems may fail for the following reasons:
1. Insufficient context: A short sentence may not contain enough information to identify the intended sense.
2. Domain-specific meanings: A word may have a specialized meaning in a legal, medical, or technical document.
3. Unseen senses: The required meaning may be missing from the lexical database or model's training data.
4. Figurative language: Metaphors, idioms, and unusual expressions may confuse the system.
5. Multilingual variation: Regional language use, transliteration, and code-switching can introduce additional ambiguity.
Therefore, successful word sense disambiguation requires both lexical knowledge and contextual understanding.

**Q.2 A company is building an NLP system to classify word senses using Naïve Bayes.**

(a) Briefly explain how Naïve Bayes can be applied to Word Sense Disambiguation (WSD).
Word Sense Disambiguation (WSD) is the process of identifying the intended meaning of an ambiguous word based on its context.

The Naïve Bayes classifier uses Bayes' theorem to calculate the probability of each possible word sense given the surrounding words.

Bayes' theorem:
\[
P(S\mid C)=\frac{P(C\mid S)P(S)}{P(C)}
\]

Where:
- \(S\) = a possible sense of the ambiguous word.
- \(C\) = the context surrounding the word.
- \(P(S\mid C)\) = probability of the sense given the context.
- \(P(C\mid S)\) = probability of the context given the sense.
- \(P(S)\) = prior probability of the sense.
- \(P(C)\) = probability of the context.
Working procedure:
1. Collect a dataset of sentences labeled with the correct word senses.
2. Extract contextual features, such as nearby words and parts of speech.
3. Calculate the prior probability of each sense and the likelihood of the observed features.
4. Estimate the posterior probability for every candidate sense.
5. Select the sense with the highest posterior probability.
Example:

Consider the ambiguous word “bank.”
- Sentence 1: “She deposited money in the bank.”
- Sentence 2: “The children sat beside the river bank.”
The words “deposited” and “money” support the financial sense, while “river” supports the river-edge sense.

Naïve Bayes learns these contextual associations from labeled examples and predicts the most probable sense for a new sentence.
(b) What are its key limitations compared to Lesk’s knowledge-based approach?
Naïve Bayes	Lesk's algorithm
Uses a probabilistic, usually supervised approach.	Uses dictionary definitions and glosses to compare meanings.
Requires labeled training data.	Can work without labeled training examples if lexical definitions are available.
Assumes contextual features are conditionally independent given the sense.	Relies on overlap between the context and candidate sense definitions.
May perform poorly on rare or unseen senses.	May fail when definitions have little word overlap with the sentence.
Can learn domain-specific patterns from suitable training data.	Depends on the coverage and quality of the lexical resource.
Limitations of Naïve Bayes:
1. Independence assumption: Words in a sentence are related, but the model treats their features as conditionally independent.
2. Data requirement: It needs sufficient labeled examples for each sense.
3. Sparse data: Rare words and uncommon senses may not have enough training examples.
4. Limited contextual understanding: It may miss long-range relationships and subtle meanings.
Lesk's algorithm instead selects a sense by comparing the words in a sentence's context with the definitions of candidate senses. It is useful when labeled data is unavailable, but its accuracy depends heavily on definition overlap.
(c) Propose one improvement using a semi-supervised method such as Yarowsky's algorithm.
Yarowsky's algorithm improves WSD by combining a small amount of labeled data with a large amount of unlabeled text.

Proposed improvement:
1. Begin with a small set of sentences labeled with the correct senses.
2. Train an initial Naïve Bayes classifier on the labeled dataset.
3. Use the classifier to predict senses for unlabeled sentences.
4. Select high-confidence predictions and add them to the labeled dataset as pseudo-labeled examples.
5. Retrain the classifier using the expanded dataset.
6. Repeat the process until performance stabilizes or a stopping condition is reached.
Example:

Initial labeled examples:
- “She deposited money at the bank.” — Financial sense.
- “The river overflowed its bank.” — River-edge sense.
The classifier then processes thousands of unlabeled sentences. A sentence such as “The boat reached the bank of the river” may be assigned the river-edge sense with high confidence and added to the training data.

Advantages:
- Reduces dependence on manually labeled data.
- Uses large amounts of available text.
- Can improve classification when high-confidence predictions are reliable.
Precaution: Incorrect pseudo-labels can reinforce errors. Confidence thresholds, balanced training examples, and validation on a separate labeled dataset help reduce this problem.

**Q.3 Consider the sentence: “John met Alex after he finished the project.”**

(a) Identify the anaphora and discuss why reference resolution is difficult here.
Anaphora occurs when an expression refers to an entity mentioned earlier in the discourse.

In the given sentence, the pronoun “he” is the anaphor, and the possible antecedents are John and Alex.

The sentence is ambiguous because both names refer to people who could grammatically be associated with the pronoun.
- Interpretation 1: John finished the project before meeting Alex.
- Interpretation 2: Alex finished the project before John met Alex.
Reference resolution is difficult because:
1. Both candidates are grammatically plausible.
2. The sentence does not provide enough semantic information to establish who completed the project.
3. Real-world knowledge may suggest possible interpretations but cannot determine the correct one reliably.
4. Additional sentences or discourse context may be necessary to resolve the ambiguity.
Thus, syntactic information alone may not uniquely identify the antecedent.
(b) How would Hobbs' algorithm attempt to resolve “he”?
Hobbs' algorithm is a syntax-based algorithm for resolving pronouns by examining parse trees and searching for suitable antecedents in a prescribed order.

The basic procedure is:
1. Parse the sentence into a syntactic tree.
2. Locate the pronoun “he” in the parse tree.
3. Search the relevant syntactic structures for noun phrases that precede the pronoun.
4. Check candidate noun phrases for agreement in person, number, and gender.
5. Select the first suitable candidate according to the algorithm's tree-traversal rules.

However, Hobbs' algorithm does not guarantee the intended interpretation. It primarily uses syntactic structure and agreement constraints; the sentence lacks enough information to determine definitively who finished the project.
(c) Discuss one real-world scenario where syntactic constraints alone fail to ensure coherence.
Scenario: Workplace conversation

Consider the sentence:

“Priya informed Neha that she had received the promotion.”

The pronoun “she” may refer to either Priya or Neha. Both candidates are grammatically compatible with the pronoun.

If the surrounding conversation explains that Neha had been waiting for the promotion, then Neha is likely the intended antecedent. Without this context, a syntax-based system may select the wrong person.

This demonstrates that syntactic constraints alone cannot ensure coherent reference resolution. Systems may also need discourse context, semantic relationships, speaker intentions, and real-world knowledge.

Other examples include pronouns in news reports, medical records, and legal documents, where assigning a pronoun to the wrong person could change the meaning substantially.
